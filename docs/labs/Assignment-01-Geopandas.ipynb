## 1. Import the packages

## 2. Load the first species table

Start with corn snake, *Pantherophis guttatus*. The first goal is only to read the CSV file and inspect the first few rows.

Complete the filename in the first line, then run the cell.

```python
file = "EasternSnakes/Pguttatus_coords.csv"
pguttatus_csv = file

pguttatus_df = gpd.read_file(pguttatus_csv)
pguttatus_df.head(pguttatus_df)
```

## 3. Convert the first species to a GeoDataFrame

The table has longitude and latitude columns, but those columns are not yet geometry. Use `gpd.points_from_xy()` to make point geometry.

Fill in the longitude and latitude column names. Look back at the output from the previous cell if you need a reminder.

```python
pguttatus_gdf = gpd.GeoDataFrame(
    pguttatus_df,
    geometry=gpd.points_from_xy(pguttatus_df["Longitude"], pguttatus_df["Latitude"]),
    crs="EPSG:4326",
)

pguttatus_gdf["Species"] = "Pantherophis guttatus"
pguttatus_gdf.head()
```

## 4. Map the first species

```python
pguttatus_gdf.explore()
```

## 14. Project the data before measuring area

Longitude and latitude are measured in degrees. Area measured in degrees is not easy to interpret, so project the data to `EPSG:5070` before calculating polygon areas.

```python
combined_projected = combined_snakes.to_crs("EPSG:5070")
combined_projected.crs
```

## 15. Make a convex hull for each species

A convex hull is a simple polygon drawn around a set of points. It is not the true species range, but it is a useful first practice tool.

This cell does three things: groups records by species, turns each group into a convex hull, and maps the hulls.

```python
hulls_projected = combined_projected.dissolve(by="Species")
hulls_projected = hulls_projected.reset_index()

hulls_projected["geometry"] = hulls_projected.geometry.convex_hull

hulls = hulls_projected.to_crs(combined_snakes.crs)
hulls.explore(column="Species", cmap="rainbow")
```

## 16. Calculate convex hull area

`hulls_projected.area` gives area in square meters because `EPSG:5070` uses meters. Divide by 1,000,000 to convert square meters to square kilometers.

Fill in the missing column name.

```python
hulls["convex_hull_area_km2"] = hulls_projected.area / 1000000

hulls[["Species", "convex_hull_area_km2"]]
```
